```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd

file_path = "/content/drive/MyDrive/university_final_dataset.csv"
df = pd.read_csv(file_path)

print("Shape before:", df.shape)
print("Total null values before:", df.isnull().sum().sum())

row_threshold = df.shape[1] * 0.95
df = df.dropna(axis=0, thresh=row_threshold)

num_cols = df.select_dtypes(include='number').columns
df[num_cols] = df[num_cols].fillna(df[num_cols].mean())

text_cols = df.select_dtypes(include='object').columns
for col in text_cols:
    if not df[col].mode().empty:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Shape after:", df.shape)
print("Total null values after:", df.isnull().sum().sum())

output_path = "/content/drive/MyDrive/cleaned_dataset_rows_5percent.csv"
df.to_csv(output_path, index=False)

print("Saved successfully!")
```

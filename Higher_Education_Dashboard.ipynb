```python
import pandas as pd

qs = pd.read_csv('/content/drive/MyDrive/qs-world-rankings-2025.csv')
the = pd.read_csv('/content/drive/MyDrive/THE World University Rankings 2016-2026.csv')
```

```python
qs = qs.drop_duplicates()
the = the.drop_duplicates()
```

```python
qs.isnull().sum()
the.isnull().sum()
```

```python
qs['Institution Name'] = qs['Institution Name'].str.strip().str.lower()
the['Name'] = the['Name'].str.strip().str.lower()

qs['Location'] = qs['Location'].str.strip().str.lower()
the['Country'] = the['Country'].str.strip().str.lower()
```

```python
merged = pd.merge(
    qs,
    the,
    left_on=['Institution Name', 'Location'],
    right_on=['Name', 'Country'],
    how='outer'
)
```

```python
merged.head()
```

```python
merged.info()
```

```python
df = merged.copy()
```

```python
df.info()
```

```python
df['Student_Population_Cleaned'] = (
    df['Student Population']
    .astype(str)
    .str.replace(',', '', regex=False)
)

df['Student_Population_Cleaned'] = pd.to_numeric(
    df['Student_Population_Cleaned'],
    errors='coerce'
)
```

```python
df[['Student Population', 'Student_Population_Cleaned']].head()
```

```python
df['International_Students_y_Cleaned'] = (
    df['International Students_y']
    .astype(str)
    .str.replace('%', '', regex=False)
    .str.replace(',', '', regex=False)
)

df['International_Students_y_Cleaned'] = pd.to_numeric(
    df['International_Students_y_Cleaned'],
    errors='coerce'
)
```

```python
df[['International Students_y', 'International_Students_y_Cleaned']].head()
```

```python
df['QS Overall Score'] = pd.to_numeric(
    df['QS Overall Score'],
    errors='coerce'
)

df['Global_Rank_Score'] = (
    df['QS Overall Score'].fillna(0) +
    df['Overall Score'].fillna(0)
) / 2
```

```python
df[['QS Overall Score',
    'Overall Score',
    'Global_Rank_Score']].head()
```

```python
df['Research_Productivity_Index'] = (
    df['Research Quality'].fillna(0) +
    df['Research Environment'].fillna(0) +
    df['Citations per Faculty'].fillna(0)
) / 3
```

```python
df[['Research Quality',
    'Research Environment',
    'Citations per Faculty',
    'Research_Productivity_Index']].head()
```

```python
df['Intl_Student_Percentage'] = df['International_Students_y_Cleaned']
```

```python
df[['Student_Population_Cleaned',
    'International_Students_y_Cleaned',
    'Global_Rank_Score',
    'Research_Productivity_Index',
    'Intl_Student_Percentage']].describe()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[['Global_Rank_Score',
    'Research_Productivity_Index',
    'Intl_Student_Percentage']] = scaler.fit_transform(
    df[['Global_Rank_Score',
        'Research_Productivity_Index',
        'Intl_Student_Percentage']]
)
```

```python
df['Performance_Index'] = (
    0.4 * df['Global_Rank_Score'] +
    0.3 * df['Research_Productivity_Index'] +
    0.3 * df['Intl_Student_Percentage']
)
```

```python
import plotly.express as px

fig = px.bar(
    df.sort_values('Performance_Index', ascending=False).head(10),
    x='Institution Name',
    y='Performance_Index',
    title='Top 10 Universities by Performance Index'
)

fig.show()
```

```python
df.sort_values('Performance_Index', ascending=False)[
    ['Institution Name', 'Name', 'Performance_Index']
].head(10)
```

```python
df['University Name'] = df['Institution Name'].fillna(df['Name'])
```

```python
import plotly.express as px

fig = px.bar(
    df.sort_values('Performance_Index', ascending=False).head(10),
    x='University Name',
    y='Performance_Index',
    title='Top 10 Universities by Performance Index'
)

fig.show()
```

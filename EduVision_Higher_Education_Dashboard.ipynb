```python
qs = pd.read_csv('/content/drive/MyDrive/Infosys Project /qs-world-rankings-2025.csv')
the = pd.read_csv('/content/drive/MyDrive/Infosys Project /THE World University Rankings 2016-2026.csv')

print("QS Dataset Shape:", qs.shape)
print("THE Dataset Shape:", the.shape)
```

```python
print("========== QS DATASET ==========")
display(qs.head())
print("\n========== THE DATASET ==========")
display(the.head())

print("\nQS Columns:")
print(qs.columns.tolist())

print("\nTHE Columns:")
print(the.columns.tolist())
```

```python
print("========== QS INFO ==========")
qs.info()

print("\n========== THE INFO ==========")
the.info()
```

```python
the = the[the['Year'] == 2025].copy()

print("THE Dataset after filtering:")
print(the.shape)

display(the.head())
```

```python
print("QS Duplicate Rows:", qs.duplicated().sum())
print("THE Duplicate Rows:", the.duplicated().sum())
```

```python
print("========== QS Missing Values ==========")
print(qs.isnull().sum())

print("\n========== THE Missing Values ==========")
print(the.isnull().sum())
```

```python
qs['Institution Name'] = qs['Institution Name'].str.strip().str.lower()
the['Name'] = the['Name'].str.strip().str.lower()

print("University names standardized successfully!")
```

```python
qs['Location Full'] = qs['Location Full'].str.strip().str.lower()
the['Country'] = the['Country'].str.strip().str.lower()

print("Country names standardized successfully!")
```

```python
common_universities = set(qs['Institution Name']) & set(the['Name'])

print("Total Universities in QS:", len(qs))
print("Total Universities in THE:", len(the))
print("Matching Universities:", len(common_universities))
```

```python
qs_only = sorted(set(qs['Institution Name']) - set(the['Name']))
the_only = sorted(set(the['Name']) - set(qs['Institution Name']))

print("Universities only in QS:", len(qs_only))
print("Universities only in THE:", len(the_only))

print("\nFirst 20 Universities only in QS:")
print(qs_only[:20])

print("\nFirst 20 Universities only in THE:")
print(the_only[:20])
```

```python
merged = pd.merge(
    qs,
    the,
    left_on=['Institution Name', 'Location Full'],
    right_on=['Name', 'Country'],
    how='inner'
)

print("Merged Dataset Shape:", merged.shape)

display(merged.head())
```

```python
merged['QS Overall Score'] = pd.to_numeric(
    merged['QS Overall Score'],
    errors='coerce'
)

merged['2025 Rank'] = pd.to_numeric(
    merged['2025 Rank'],
    errors='coerce'
)

merged['Student Population'] = (
    merged['Student Population']
    .astype(str)
    .str.replace(',', '', regex=False)
)

merged['Student Population'] = pd.to_numeric(
    merged['Student Population'],
    errors='coerce'
)

merged['International Students_y'] = (
    merged['International Students_y']
    .str.replace('%', '', regex=False)
)

merged['International Students_y'] = pd.to_numeric(
    merged['International Students_y'],
    errors='coerce'
)

print("Numeric columns cleaned successfully!")
```

```python
merged[['2025 Rank',
        'QS Overall Score',
        'Student Population',
        'International Students_y']].info()
```

```python
missing = merged.isnull().sum()

missing = missing[missing > 0].sort_values(ascending=False)

print(missing)
```

```python
merged['Research Performance Score'] = (
    merged['Research Environment'] * 0.4 +
    merged['Research Quality'] * 0.6
)

print("Research Performance Score created successfully!")

merged[['Institution Name',
        'Research Environment',
        'Research Quality',
        'Research Performance Score']].head()
```

```python
merged['Internationalization Score'] = (
    merged['International Faculty'] * 0.5 +
    merged['International Students_y'] * 0.5
)

print("Internationalization Score created successfully!")

merged[['Institution Name',
        'International Faculty',
        'International Students_y',
        'Internationalization Score']].head()
```

```python
merged['Teaching Excellence Score'] = (
    merged['Teaching'] * 0.6 +
    merged['Students to Staff Ratio'] * 0.4
)

print("Teaching Excellence Score created successfully!")

merged[['Institution Name',
        'Teaching',
        'Students to Staff Ratio',
        'Teaching Excellence Score']].head()
```

```python
max_ratio = merged['Students to Staff Ratio'].max()

merged['Staff Availability Score'] = (
    (max_ratio - merged['Students to Staff Ratio']) / max_ratio
) * 100

print("Staff Availability Score created successfully!")

merged[['Institution Name',
        'Students to Staff Ratio',
        'Staff Availability Score']].head()
```

```python
merged['Teaching Excellence Score'] = (
    merged['Teaching'] * 0.6 +
    merged['Staff Availability Score'] * 0.4
)

print("Teaching Excellence Score updated successfully!")

merged[['Institution Name',
        'Teaching',
        'Staff Availability Score',
        'Teaching Excellence Score']].head()
```

```python
merged['Industry Collaboration Score'] = merged['Industry Impact']

print("Industry Collaboration Score created successfully!")

merged[['Institution Name',
        'Industry Impact',
        'Industry Collaboration Score']].head()
```

```python
merged['Overall Performance Index'] = (
    merged['Research Performance Score'] * 0.35 +
    merged['Teaching Excellence Score'] * 0.35 +
    merged['Internationalization Score'] * 0.15 +
    merged['Industry Collaboration Score'] * 0.15
)

print("Overall Performance Index created successfully!")

merged[['Institution Name',
        'Research Performance Score',
        'Teaching Excellence Score',
        'Internationalization Score',
        'Industry Collaboration Score',
        'Overall Performance Index']].head()
```

```python
kpi_columns = [
    'Institution Name',
    'Location Full',
    'Research Performance Score',
    'Internationalization Score',
    'Staff Availability Score',
    'Teaching Excellence Score',
    'Industry Collaboration Score',
    'Overall Performance Index'
]

display(merged[kpi_columns].head())

print("\nTotal KPI Columns:", len(kpi_columns))
```

```python
output_file = '/content/University_Performance_Dashboard_Data.csv'

merged.to_csv(output_file, index=False)

print("Dataset exported successfully!")
print("File saved as:", output_file)
```
